Named Entity Recognition (NER)


```python
from transformers import pipeline

# Initialize the NER pipeline
ner_pipeline = pipeline("ner", model="dslim/bert-base-NER")
print("NER pipeline initialized successfully.")
```

```python
sample_title_for_ner = df.iloc[10466]["title"]
print("--- Sample Title ---")
print(sample_title_for_ner)

# Perform NER on the title
ner_results_title = ner_pipeline(sample_title_for_ner)

print("\n--- NER Results for Title ---")
if ner_results_title:
    for entity in ner_results_title:
        print(f"Entity: {entity['word']}, Type: {entity['entity']}, Score: {entity['score']:.4f}")
else:
    print("No named entities found in the title.")
```

```python
from sentence_transformers import SentenceTransformer
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import faiss

embedding_model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

summarization_tokenizer = AutoTokenizer.from_pretrained("facebook/bart-large-cnn")
summarization_model = AutoModelForSeq2SeqLM.from_pretrained("facebook/bart-large-cnn")

def search_paper(query, k=5):
    query_embedding = embedding_model.encode([query])
    faiss.normalize_L2(query_embedding)
    D, I = index.search(query_embedding, k)
    return D, I

def summarizer_search(query, k=5):
    D, I = search_paper(query, k)
    for score, idx in zip(D[0], I[0]):
        paper_title = df.iloc[idx]['title']
        paper_abstract = df.iloc[idx]['abstract']

        # Generate summary for the abstract using the summarization model
        inputs = summarization_tokenizer([paper_abstract], max_length=1024, return_tensors='pt', truncation=True)
        summary_ids = summarization_model.generate(
            inputs['input_ids'],
            num_beams=4,
            max_length=150,
            min_length=30,
            early_stopping=True
        )
        generated_summary = summarization_tokenizer.decode(summary_ids[0], skip_special_tokens=True)

        # Extract keywords using KeyBERT
        keywords = kw_model.extract_keywords(paper_abstract, keyphrase_ngram_range=(1, 3), stop_words='english')
        keywords_formatted = ", ".join([f'{word} ({score:.2f})' for word, score in keywords])

        # Perform NER on the abstract
        ner_results = ner_pipeline(paper_abstract)
        entities_formatted = []
        for entity in ner_results:
            # Filter out generic words or very short entities that are not very informative
            if len(entity['word'].strip()) > 1 and entity['entity'] not in ['O', 'MISC']:
                entities_formatted.append(f"{entity['word']} ({entity['entity']})")
        entities_formatted_str = ", ".join(list(set(entities_formatted))) # Use set to remove duplicates

        print(f"\n--- Search Result (Score: {score:.4f}) ")
        print(f"Title: {paper_title}")
        print(f"Abstract: {paper_abstract[:500]}")
        print(f"Summary: {generated_summary}")
        print(f"Keywords: {keywords_formatted}")
        print(f"Named Entities: {entities_formatted_str if entities_formatted_str else 'None found'}")

summarizer_search("deep learning for medical image analysis", k=3)
```

```python
from sentence_transformers import SentenceTransformer
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import faiss

embedding_model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")


summarization_tokenizer = AutoTokenizer.from_pretrained("facebook/bart-large-cnn")
summarization_model = AutoModelForSeq2SeqLM.from_pretrained("facebook/bart-large-cnn")

def search_paper(query, k=5):
    query_embedding = embedding_model.encode([query])
    faiss.normalize_L2(query_embedding)
    D, I = index.search(query_embedding, k)
    return D, I

def summarizer_search(query, k=5, query_keyword_str=None):
    D, I = search_paper(query, k)
    for score, idx in zip(D[0], I[0]):
        paper_title = df.iloc[idx]['title']
        paper_abstract = df.iloc[idx]['abstract']

        # Generate summary for the abstract using the summarization model
        inputs = summarization_tokenizer([paper_abstract], max_length=1024, return_tensors='pt', truncation=True)
        summary_ids = summarization_model.generate(
            inputs['input_ids'],
            num_beams=4,
            max_length=150,
            min_length=30,
            early_stopping=True
        )
        generated_summary = summarization_tokenizer.decode(summary_ids[0], skip_special_tokens=True)

        # Extract keywords using KeyBERT
        paper_keywords = kw_model.extract_keywords(paper_abstract, keyphrase_ngram_range=(1, 3), stop_words='english')
        keywords_formatted = ", ".join([f'{word} ({s_score:.2f})' for word, s_score in paper_keywords])

        # Perform NER on the abstract
        ner_results = ner_pipeline(paper_abstract)
        entities_formatted = []
        for entity in ner_results:
            # Filtering out generic words
            if len(entity['word'].strip()) > 1 and entity['entity'] not in ['O', 'MISC']:
                entities_formatted.append(f"{entity['word']} ({entity['entity']})")
        entities_formatted_str = ", ".join(list(set(entities_formatted))) # Use set to remove duplicates

        print(f"\n--- Search Result (Score: {score:.4f}) ")
        print(f"Title: {paper_title}")
        print(f"Abstract: {paper_abstract[:500]}...")
        print(f"Summary: {generated_summary}")
        print(f"Keywords: {keywords_formatted}")
        print(f"Named Entities: {entities_formatted_str if entities_formatted_str else 'None found'}")

        # Explanation Logic for XAI
        if query_keyword_str is not None:
            paper_keyword_str = [k[0] for k in paper_keywords]
            common_keywords = list(set(query_keyword_str) & set(paper_keyword_str))

            # Determine confidence based on the score
            if score >= 0.80:
                confidence = "high"
            elif score >= 0.60:
                confidence = "medium"
            else:
                confidence = "low"

            print(f"\n--- Explanation for Recommendation (Confidence: {confidence}) ---")
            if common_keywords:
                print(
                    f"This paper was retrieved because it shares "
                    f"the concepts: {', '.join(common_keywords)} "
                    f"with your query."
                )
            else:
                print(
                    "This paper was retrieved due to overall semantic similarity (no direct keyword matches)."
                )
            print(f"Matching Concepts with Query: {', '.join(common_keywords) if common_keywords else 'None'}")
        print("="*60) # Separator for next result

summarizer_search("deep learning for medical image analysis", k=3, query_keyword_str=query_keyword_str)
```

```python
@tool
def search_and_summarize(query: str, k: int = 5):
    """
    Search research papers from the FAISS database,
    retrieve the top-k most similar papers,
    summarize each paper using BART,
    and return the results as structured data.
    """

    query_embedding = model.encode([query])

    faiss.normalize_L2(query_embedding)


    D, I = index.search(query_embedding, k)

    papers = []

    for rank, (score, idx) in enumerate(zip(D[0], I[0]), start=1):

        paper = df.iloc[idx]

        summary = summarizer(
            paper["abstract"],
            max_length=120,
            min_length=40,
            do_sample=False
        )[0]["summary_text"]


        paper_data = {
            "rank": rank,
            "similarity": round(float(score), 4),
            "title": paper["title"],
            "abstract": paper["abstract"],
            "summary": summary
        }

        papers.append(paper_data)

    return papers
```

```python
@tool
def extract_keywords(text: str, top_n: int = 5) -> str:
    """
    Extract the most important keywords from the given text
    using the KeyBERT model.
    """


    keywords = kw_model.extract_keywords(
        text,
        keyphrase_ngram_range=(1, 2),
        stop_words="english",
        top_n=top_n
    )

    result = "Top Keywords:\n\n"

    for rank, (keyword, score) in enumerate(keywords, start=1):

        result += (
            f"{rank}. {keyword} "
            f"(Relevance Score: {round(score, 4)})\n"
        )

    return result
```

```python
from langchain_core.tools import tool

@tool
def compare_papers(
    paper1: str,
    paper2: str
) -> str:
    """
    Compare two research papers based on their abstracts.
    The tool retrieves the closest matching papers from the FAISS database
    and uses the LLM to compare them.
    """


    embedding1 = model.encode([paper1])

    faiss.normalize_L2(embedding1)

    D1, I1 = index.search(embedding1, 1)

    first_paper = df.iloc[I1[0][0]]


    embedding2 = model.encode([paper2])

    faiss.normalize_L2(embedding2)

    D2, I2 = index.search(embedding2, 1)

    second_paper = df.iloc[I2[0][0]]


    comparison_prompt = f"""
Compare the following two research papers.

Paper 1

Title:
{first_paper['title']}

Abstract:
{first_paper['abstract']}


Paper 2

Title:
{second_paper['title']}

Abstract:
{second_paper['abstract']}


Compare them based on:

1. Research Objective

2. Methodology

3. Key Contributions

4. Advantages

5. Limitations

6. Applications

Present the comparison in a clear table.
"""


    response = llm.invoke(comparison_prompt)

    return response.content
```
